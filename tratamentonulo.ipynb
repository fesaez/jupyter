Tratamento nulo
Durante a limpeza de tratamento dos dados, é comum lidar com dados faltantes: A biblioteca Pandas oferece funcionalidades para encontrar e tratar esses valores nulos.
ISNA - Para verificar a presença de dados nulos em uma DataFrame ou Series, podemos usar o método isna(). Ele verifica cada elemento e retorna TRUE se o valor for nulo.
Valores booleanos pode ser identificados como números interiros, onde TRUE é equivalente a 1 e FALSE a 0. Podemos usar o método de agregação sum junto com isna() para contar a quantidade de valores nulos em um DataFrame.

fillna: o método fillna() substitui os valores nulos por um valor escolhido pelo usuário.

```python
import numpy as np
```

```python
dados = {
    'id_cliente': [101, 102, 103,104],
    'saldo': [np.nan, 6500, 7000, 800],
    'tipo_de_conta': ['corrente', 'poupanca', np.nan, 'corrente']
}
```

```python
df = pd.DataFrame(dados)
df.isna()
```
