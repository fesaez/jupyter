```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
import pandas as pd

# Carregue seus dados no DataFrame 'df' aqui.
# Exemplo: df = pd.read_csv('seu_arquivo.csv')
# Para demonstração, vamos criar um df simples:
data = {'Ano': [2020, 2020, 2021, 2021, 2022, 2022],
        'Consumo': [100, 150, 120, 180, 130, 190],
        'Mês': [1, 2, 3, 4, 5, 6]}
df = pd.DataFrame(data)

consumo_anual = df.groupby('Ano')['Consumo'].sum().reset_index()

plt.figure(figsize=(10,5))
sns.lineplot(data=consumo_anual, x='Ano', y='Consumo', marker='o')
plt.title('Consumo Anual de Energia')
plt.xlabel('Ano')
plt.ylabel('Consumo')
plt.grid(True)
plt.show()
```

```python
consumo_mensal = df.groupby('Mês')['Consumo'].mean().reset_index()

plt.figure(figsize=(12,5))
sns.barplot(data=consumo_mensal, x='Mês', y='Consumo')
plt.title('Consumo Médio por Mês')
plt.xlabel('Mês')
plt.ylabel('Consumo Médio')
plt.xticks(rotation=45)
plt.show()
```

```python
# Adiciona a coluna 'Valor Fatura' ao DataFrame df para demonstração
# Em um cenário real, você carregaria esta coluna de seus dados ou a calcularia.
# Certifique-se de que o número de elementos corresponda ao número de linhas em df.
if 'Valor Fatura' not in df.columns:
    # Usando valores de exemplo que correspondam ao número de linhas do df atual (6 linhas)
    df['Valor Fatura'] = [20, 25, 22, 28, 23, 30]

correlacao = df[['Consumo', 'Valor Fatura']].corr()

plt.figure(figsize=(6,4))
sns.heatmap(correlacao, annot=True, cmap='Blues')
plt.title('Correlação entre Consumo e Valor da Fatura')
plt.show()
```
