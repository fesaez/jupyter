```python
# Escreva uma função chamada dividir_numeros - recebe dois números como argumentos e retorna
# o resultado da divisão do primeiro número pelo segundo.
# Se o segundo número for zero, a função deve lançar uma exceção ValueError com a mensagem "O divisor não pode ser zero".
#Crie um bloco try e except para chamar a função dividir_numeros com diferentes conjuntos de números,
#incluindo um caso em que o divisor é zero. Capture a exceção ValueError e imprima a mensagem de erro.
```

```python
def dividir_numeros(a, b):
  if b == 0:
    raise ValueError("O divisor não pode ser zero")
  return a / b

#Teste a função dividir_numeros com diferentes conjuntos de núemros
```

```python
try:
  resultado = dividir_numeros(10, 2)
  print (f"Resultado da divisao: {resultado}")

  resultado = dividir_numeros(8, 0) #Isso deve lançar uma exceção
  print (f"Resultado da divisão: {resultado}")
except ValueError as e:
  print (f"Erro {e}")
```
