```python
#Define uma classe conta
#Métodos: inicialização, depósito, saque
#atributos: número de contas e saldo
#Métodos: executa ações - depositar e sacar
class Conta:
  def __init__(self, numero_da_conta, saldo):
    self.numero_da_conta = numero_da_conta
    self.saldo = saldo
  def depositar(self, valor):
    self.saldo += valor
  def sacar(self, valor):
    self.saldo -= valor

```

```python
class ContaPoupanca (Conta):
  def calcular_rendimento(self):
    self.saldo += self.saldo*0.05
cp = ContaPoupanca(numero_da_conta = 123456, saldo = 200)
```

```python
cp.depositar(100)
cp.saldo
```

```python
cp.sacar(50)
cp.saldo
```

```python
cp.calcular_rendimento()
cp.saldo
```
