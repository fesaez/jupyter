```python
class ContaPoupanca(conta):
  def sacar(self, valor):
    if self.saldo - valor > 0:
      self.saldo -= valor
```

```python
class ContaCorrente(conta):
  limite = 1000

  def sacar(self, valor):
    if self.saldo + self.limite - valor > 0:
      self.saldo -= valor
```

```python
cc = ContaCorrente(numero_da_conta=54321, saldo=200)
```

```python
cc.sacar(500)
print(cc.saldo)
```
