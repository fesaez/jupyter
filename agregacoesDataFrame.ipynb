```python
import pandas as pd
df_nomes = pd.DataFrame ({
    'id_cliente': [1, 2, 3, 4],
    'nome': ['joao', 'jose', 'maria', 'ana']
})
df_nomes
```

```python
df_vendas = pd.DataFrame({
    'id_cliente': [1, 2, 3, 4],
    'quantidade_vendadida': [10, 150, 200, 250]
})
df_vendas
```

```python
df_nome_venda = pd.merge(df_nomes, df_vendas, on='id_cliente')
df_nome_venda
```

```python
df_nome_venda_nova = pd.DataFrame({
    'id_cliente':[5],
    'nome': ['tiago'],
    'quantidade_vendido':[100]
})
pd.concat([df_nome_venda, df_nome_venda_nova])
```
