```python
import pandas as pd
from sklearn.model_selection import train_test_split

class DataPrep:
    def __init__(self, df):
        self.df = df.copy()  # Usa uma cópia para evitar modificar o df_fraude original diretamente

    def preparar_dados(self):
        # Para demonstração com LogisticRegression, vamos criar uma variável alvo binária
        # usando 'median_house_value'. Casas acima do valor mediano serão 1.
        if 'median_house_value' in self.df.columns:
            self.df['high_value_house'] = (self.df['median_house_value'] > self.df['median_house_value'].median()).astype(int)
            target_column = 'high_value_house'
            # Features (X) são todas as colunas exceto 'median_house_value' e a nova variável alvo 'high_value_house'
            features = self.df.drop(columns=['median_house_value', target_column])
            target = self.df[target_column]
        else:
            # Se 'median_house_value' não estiver presente, ou para outro conjunto de dados,
            # fallback para DataFrames vazios ou lançar um erro.
            # Por enquanto, assumimos que 'median_house_value' existe com base no estado atual do df_fraude.
            print("Aviso: 'median_house_value' não encontrado. Retornando DataFrames vazios.")
            return pd.DataFrame(), pd.DataFrame()

        df_treino_X, df_teste_X, df_treino_y, df_teste_y = train_test_split(
            features, target, test_size=0.2, random_state=42
        )

        # Recombina X e y em DataFrames para manter consistência com a estrutura original do código do usuário
        df_treino = df_treino_X.copy()
        df_treino[target_column] = df_treino_y

        df_teste = df_teste_X.copy()
        df_teste[target_column] = df_teste_y

        return df_treino, df_teste

dp = DataPrep(df_fraude)

df_treino, df_teste = dp.preparar_dados()
```

```python
from sklearn.linear_model import LogisticRegression

# Aumentando o número máximo de iterações para evitar o ConvergenceWarning
clf = LogisticRegression(max_iter=1000)

# Using 'high_value_house' as the target column, as created in DataPrep
X_treino = df_treino.drop(columns='high_value_house')
y_treino = df_treino['high_value_house']

X_teste = df_teste.drop(columns='high_value_house')
y_teste = df_teste['high_value_house']
```

```python
import numpy as np
from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import make_scorer, f1_score, roc_auc_score, classification_report

# ---------------------------
# 1. Usando os dados já separados e preparados de etapas anteriores
#    (df_treino, df_teste, high_value_house)
# ---------------------------
X_train = X_treino
y_train = y_treino
X_test = X_teste
y_test = y_teste

# ---------------------------
# 2. Pipeline (sem leakage)
# ---------------------------
pipeline = Pipeline([
    ("scaler", StandardScaler()),

    # Feature selection acontece DENTRO da CV (evita leakage)
    ("feature_selection", SelectKBest(score_func=f_classif)),

    ("model", LogisticRegression(
        class_weight="balanced",  # ajuda no desbalanceamento
        max_iter=1000
    ))
])

# ---------------------------
# 3. Validação cruzada correta
# ---------------------------
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

# ---------------------------
# 4. Métricas apropriadas
# ---------------------------
scoring = {
    "f1": make_scorer(f1_score),
    "roc_auc": "roc_auc"
}

# ---------------------------
# 5. Grid Search (inclui seleção de variáveis)
# ---------------------------
param_grid = {
    "feature_selection__k": [5, 10, 20, "all"],
    "model__C": [0.01, 0.1, 1, 10]
}

grid = GridSearchCV(
    pipeline,
    param_grid=param_grid,
    cv=cv,
    scoring="f1",   # escolha principal da otimização
    n_jobs=-1
)

# ---------------------------
# 6. Treinamento
# ---------------------------
grid.fit(X_train, y_train)

print("Melhores parâmetros:", grid.best_params_)
print("Melhor F1 CV:", grid.best_score_)

# ---------------------------
# 7. Avaliação final (SÓ UMA VEZ no teste)
# ---------------------------
best_model = grid.best_estimator_
y_pred = best_model.predict(X_test)

print("\nRelatório no teste:")
print(classification_report(y_test, y_pred))
```
